```python
import zipfile

with zipfile.ZipFile('/content/archive.zip', 'r') as zip_ref:
    zip_ref.extractall('/content/dataset')

print("Extracted to /content/dataset")

!find /content/dataset -maxdepth 3
```

```python
yaml_content = """path: /content/dataset/dataset
train: images/train
val: images/val
test: images/test

names:
  0: Knife
  1: Handgun
"""

with open('/content/dataset/dataset/data.yaml', 'w') as f:
    f.write(yaml_content)

print(yaml_content)
```

```python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')

results = model.train(
    data='/content/dataset/dataset/data.yaml',
    epochs=30,
    imgsz=416,          # smaller image size = faster loading + training
    batch=32,
    workers=2,
    fraction=0.25,
    patience=8,
    project='/content/drive/MyDrive/weapon-detection/runs',
    name='yolov8n_weapons'
)
```
